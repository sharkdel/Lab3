Добавьте:

Формулы:

### Задание 11. Работа с датами

1. Преобразуйте `order_date` в тип `datetime`.
2. Создайте столбцы:
   - `month`
   - `day_of_week`
3. Посчитайте количество заказов по дням недели.

````python
# ваш код
df.dtypes
# 1. Преобразуйте `order_date` в тип `datetime`.
df["order_date"] = pd.to_datetime(df["order_date"])
df["order_date"].dtypes

# 2. Создайте столбцы:
df["month"] = df["order_date"].dt.month
df["day_of_week"] = df["order_date"].dt.day_name()

# 3. Посчитайте количество заказов по дням недели.
print(df.groupby("day_of_week").agg(Колво_заказов_по_дням= ("order_id", "count")))
df["day_of_week"].value_counts()
df.groupby("day_of_week")["order_id"].count()

# 4. Посчитайте суммарную выручку по месяцам.
df["month"].value_counts() # кол-во заказов за месяц
df.groupby("month")["total"].sum().round(2)
print(df.groupby("month").agg(Суммарная_выручка= ("total", "sum")).round(2))
````

### Задание 14. Свой вопрос к данным

Сформулируйте вопрос, выполните расчёт и напишите короткий вывод.


````python
df.head(2)
````

````python
# ваш код
# В каком городе чаще покупают электронику
df[df["category"] == "Красота"].groupby("city").size().reset_index(name= "Колличество покупок").sort_values(
    by= "Колличество покупок", ascending=False)

# Какие товараы пользуются спросом в Казани
df[df["city"] == "Казань"].groupby("category").size().reset_index(name= "Колличество покупок").sort_values(by= "Колличество покупок", ascending=False)

# Где чаще используют сбп
df[df["payment_method"] == "СБП"].groupby("city").size().reset_index(name= "СБП").sort_values(by= "СБП", ascending=False)

#Сортировка заказов по способам оплаты
df.groupby(["city", "payment_method"]).size().unstack(fill_value=0).sort_values(by= "СБП", ascending=False)

# Оплата по СБП отностительно других способов оплаты
# Переводим условие в True/False (1 и 0) и суммируем их по городам
df.groupby("city").agg(
    СБП= ("payment_method", lambda x: (x == "СБП").sum()),
    Остальные_способы= ("payment_method", lambda x: (x != "СБП").sum()),
).sort_values("СБП", ascending=False)
````
